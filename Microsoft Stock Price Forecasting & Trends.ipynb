```python
df = pd.read_csv("Microsoft_Stock.csv", low_memory=False)
df["Date"] = pd.to_datetime(df["Date"])
df = df.sort_values("Date").reset_index(drop=True)
```

```python
df.head()
```

```python
df.info()
```

```python
plt.figure(figsize=(12, 6))
plt.plot(df["Date"], df["Close"], label="Closing Price", color="royalblue")
plt.title(
    "Microsoft Stock Closing Prices Over Time", fontsize=14, fontweight="bold"
)
plt.xlabel("Date", fontsize=12)
plt.ylabel("Price (USD)", fontsize=12)
plt.legend()
plt.show()
```

```python
df["MA_50"] = df["Close"].rolling(window=50).mean()
df["MA_200"] = df["Close"].rolling(window=200).mean()
```

```python
plt.figure(figsize=(12, 6))
plt.plot(
    df["Date"], df["Close"], label="Close Price", color="lightgray", alpha=0.7
)
plt.plot(df["Date"], df["MA_50"], label="50-Day Moving Average", color="orange")
plt.plot(
    df["Date"], df["MA_200"], label="200-Day Moving Average", color="crimson"
)
plt.title(
    "Microsoft Stock Price with Moving Averages", fontsize=14, fontweight="bold"
)
plt.xlabel("Date", fontsize=12)
plt.ylabel("Price (USD)", fontsize=12)
plt.legend()
plt.show()
```
